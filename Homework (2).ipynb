```python
from transformers import AutoTokenizer
from openai import OpenAI
import tiktoken

# Инициализация API (если нужно)
qween = OpenAI(
    base_url="https://openrouter.ai",
    api_key="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"
)

# Текст запроса
text = "Здравствуйте, у меня не работает воспроизведение фильма на телевизоре Samsung. Подписка оплачена, но при нажатии на кнопку Play экран остается черным. Перезагрузка приложения не помогла. Что делать?"

# Цены за 1 токен
priceQwenInput = 0.10 / 1000000
priceYandexInput = 200 / 1000000

print("--- РАСЧЕТ ДЛЯ QWEN (ЧЕРЕЗ ТОКЕНИЗАТОР OPENAI ДЛЯ ПРИМЕРА) ---")

tokenizer_openai = tiktoken.encoding_for_model("gpt-4o-mini")
token_idsQ = tokenizer_openai.encode(text)

print(f"Количество токенов для Qwen (примерно): {len(token_idsQ)}")
# Декодируем родным для этих ID токенизатором tiktoken
tokensQ = [tokenizer_openai.decode([tid]) for tid in token_idsQ]
print("Токены:")
print(" | ".join(tokensQ))
print(f"Стоимость ввода: {len(token_idsQ) * priceQwenInput:.8f} руб.\n")


print("--- РАСЧЕТ ДЛЯ YANDEX GPT ---")
yandex = "yandex/YandexGPT-5-Lite-8B-instruct"
tokenizer_yandex = AutoTokenizer.from_pretrained(yandex)
token_idsY = tokenizer_yandex.encode(text)

print(f"Количество токенов для yandex/YandexGPT-5-Lite-8B-instruct: {len(token_idsY)}")

tokensY = [tokenizer_yandex.decode([tid]) for tid in token_idsY]
print("Токены:")
print(" | ".join(tokensY))
print(f"Стоимость ввода: {len(token_idsY) * priceYandexInput:.8f} руб.")
```

### Подзадача 1.4. Асинхронный запрос с потоковым выводом

При синхронном запросе пользователь ждет, пока модель полностью сгенерирует ответ. Потоковый вывод (streaming) позволяет отображать текст по мере его генерации, что значительно улучшает пользовательский опыт - человек видит ответ "на лету" и может прервать генерацию, если ответ пошел не в ту сторону.

Ваша задача - переписать код из Подзадачи 1.1 для выполнения асинхронного запроса с потоковым выводом. Используйте `AsyncOpenAI` и параметр `stream=True`. Запрос: "Пользователь жалуется, что фильм останавливается каждые 5 минут и показывает буферизацию. Составь пошаговую инструкцию по решению проблемы."

**Баллы:** 4 балла.

```python
import asyncio

from openai import AsyncOpenAI

client = AsyncOpenAI(base_url="https://openrouter.ai/api/v1", api_key="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")

pormpt = "Пользователь жалуется, что фильм останавливается каждые 5 минут и показывает буферизацию. Составь пошаговую инструкцию по решению проблемы."

async def main() -> None:
    stream = await client.chat.completions.create(
        model="openai/gpt-4o-mini",
        messages=[
            {
                "role": "user",
                "content": pormpt,
            },
        ],
        stream=True,
    )
    async for chunk in stream:
        if not chunk.choices:
            continue

        print(chunk.choices[0].delta.content, end="")
    print()


await main()
```

```python
user_query = "Кратко опиши преимущества подписки Премиум в трех предложениях"

# сырой промпт, чтобы модель не рассуждала
raw_prompt = r"""<|im_start|>system
Ты — краткий ИИ-ассистент. Отвечаешь сразу по делу.<|im_end|>
<|im_start|>user
""" + user_query + r"""<|im_end|>
<|im_start|>assistant
<think></think>
"""

# 2. Токенизация текста
model_inputs = tokenizer([raw_prompt], return_tensors="pt").to(model.device)

# Убираем теги рассуждений
thought_token_ids = tokenizer.encode("<think>", add_special_tokens=False)
close_thought_ids = tokenizer.encode("</think>", add_special_tokens=False)
forbidden_tokens = [thought_token_ids, close_thought_ids]

# генерируем
with torch.no_grad():
    full_output_ids = model.generate(
        **model_inputs,
        max_new_tokens=256,            # 256 токенов точно хватит для 3 предложений
        num_beams=4,                   #
        num_return_sequences=4,        # Запрашиваем вывод всех 4 гипотез
        early_stopping=False,           # Останавливаем луч при достижении конца
        pad_token_id=tokenizer.eos_token_id,
        bad_words_ids=forbidden_tokens
    )

  # декодируем
for i, sequence_ids in enumerate(full_output_ids):

    clean_ids = sequence_ids[model_inputs.input_ids.shape[1]:]

    # декодируем чистый текст гипотезы
    hypothesis_text = tokenizer.decode(clean_ids, skip_special_tokens=True)

    # выводим на экран красивый разделитель и текст
    print(f"[Гипотеза №{i+1}]")
    print(hypothesis_text.strip())
    print("-" * 50)
```

Чем позднее гипотеза, тем ниже эти вероятности. То есть у верхних гипотез последовательность слов по вероятностям больше подходит друг к другу чем у нижних.
